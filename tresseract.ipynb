```python
import cv2
import pytesseract
import PIL 

from matplotlib import pyplot as plt    
from PIL import Image
```

```python
img = cv2.imread('./../input.png')
```

```python
if img is None:
    print("Error: Could not read the image.")
else:

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8, 4))
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.show()
```
